# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
def fib(n):
    n1, n2 = 0, 1
    fibs = []
    while n1 <= n:
        fibs.append(n1)
        n1, n2 = n2, n1+n2

    return fibs
        

x = fib(1000000)
print(x)
        
```

# Task 2
## write a program to simulate 1000 tosses of a fair coin (use np.random.binomial)
## Calculate the mean and standard deviation of that sample

```python
num_tosses = 1000
flips = np.random.binomial(n=1, p=0.5, size=num_tosses)

toss_mean = np.mean(flips)
toss_std = np.std(flips)

print(f"Mean: {toss_mean}")
print(f"STD: {toss_std}")
    
```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
import matplotlib.pyplot as plt

x = np.random.normal(loc=0, scale=5, size=200)
e = np.random.normal(loc=0, scale=1, size=200)

y = 0.5 * x + e

plt.scatter(x, y)
plt.title("Scatterplot of y=0.5*x+e")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
