# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
limit = 1_000_000
numbers = [0, 1]
first = 0
second = 1
i = 1

while numbers[i] < limit:
    numbers.append(numbers[i] + numbers[i - 1])
    i += 1

numbers.pop() # popping the last number which will be over or at the limit

print(numbers)
```

# Task 2
## write a program to simulate 1000 tosses of a fair coin (use np.random.binomial)
## Calculate the mean and standard deviation of that sample

```python
tosses = 1000
heads = np.random.binomial(1, .5, tosses) # Tossing coin

print ("Mean = " + str(np.mean(heads)))
print ("Standard Deviation = " + str(np.std(heads)))
##print (heads)
```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
x = np.random.normal(0, 5, 100) # x gaussian
e = np.random.normal(0, 1, 100) # e gaussian
y = 0.5*x + e

# Plot
plt.scatter(x, y, alpha=1)
plt.ylabel("y")
plt.xlabel("0.5*x+e")
plt.show()
```
