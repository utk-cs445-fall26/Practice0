# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
def generate_fibonacci(limit):
    fib_sequence = []
    a, b = 0, 1
    
    while a <= limit:
        fib_sequence.append(a)
        a, b = b, a + b 
        
    return fib_sequence

fib_numbers = generate_fibonacci(1000000)
print(fib_numbers)
```

# Task 2
## write a program to simulate 1000 tosses of a fair coin (use np.random.binomial)
## Calculate the mean and standard deviation of that sample

```python
import numpy as np

tosses = np.random.binomial(n=1, p=0.5, size=1000)

mean_tosses = np.mean(tosses)
std_tosses = np.std(tosses)

print(f"Mean: {mean_tosses}")
print(f"Standard Deviation: {std_tosses}")
```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
import numpy as np
import matplotlib.pyplot as plt

n = 1000
x = np.random.normal(loc=0, scale=5, size=n)
e = np.random.normal(loc=0, scale=1, size=n)
y = 0.5 * x + e

plt.scatter(x, y, alpha=0.5)
plt.title('Scatterplot of y = 0.5*x + e')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```
