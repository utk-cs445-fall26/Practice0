# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
# numbers up to less than and equal to 1000000
results = []
i = 0
res = 0
while True:
    if i == 0:
        results.append(0)
    elif i == 1:
        results.append(1)
    else:
        res = results[i-2] + results[i-1]
        if res > 1000000:
            break
        results.append(res)
    i += 1
print(results)
    
```

# Task 2
## write a program to simulate 1000 tosses of a fair coin (use np.random.binomial)
## Calculate the mean and standard deviation of that sample

```python
# flip coin 1000 times
heads = np.random.binomial(1, .5, size=1000)
# calculate mean and std deviation
mean = np.mean(heads)
std = np.std(heads)
# output
print(f"Mean: {mean}\nStandard Deviation: {std}")
```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
import matplotlib.pyplot as plt

# get distributions
x = np.random.normal(loc=0.0, scale=5.0, size=1000)
e = np.random.normal(loc=0.0, scale=1.0, size=1000)

# do equation
y = []
for i in range(1000):
    y.append(0.5 * x[i] + e[i])

# plot
plt.scatter(x, y, color='red', marker='x', alpha=1)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = 0.5x + e")
plt.show()
```
