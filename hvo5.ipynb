# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
# Task 1: produce Fibonacci numbers up to 1000000
fib = [0, 1]
while fib[-1] + fib[-2] <= 1000000:
    fib.append(fib[-1] + fib[-2])

print("Fibonacci numbers up to 1000000:")
print(fib)
```

# Task 2
## write a program to simulate 1000 tosses of a fair coin (use np.random.binomial)
## Calculate the mean and standard deviation of that sample

```python
# Task 2: simulate 1000 tosses of a fair coin and report mean and std
# Each toss: 1 trial with p=0.5 -> result is 0 (tails) or 1 (heads)
tosses = np.random.binomial(1, 0.5, size=1000)

print("Number of heads:", tosses.sum())
print("Mean:", tosses.mean())
print("Standard deviation:", tosses.std())
```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
# Task 3: scatterplot of y = 0.5*x + e
# x ~ gaussian(mean=0, std=5), e ~ gaussian(mean=0, std=1)
n = 500
x = np.random.normal(0, 5, size=n)
e = np.random.normal(0, 1, size=n)
y = 0.5 * x + e

plt.scatter(x, y, alpha=0.6, edgecolors='none')
plt.xlabel("x ~ N(0, 5)")
plt.ylabel("y = 0.5*x + e")
plt.title("Scatterplot of y = 0.5*x + e")
plt.show()
```
