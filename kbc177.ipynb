# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
a, b = 0, 1
fibonacci_numbers = []

while a <= 1000000:
    fibonacci_numbers.append(a)
    a, b = b, a + b

print("Fibonacci numbers up to 1,000,000:")
print(fibonacci_numbers)
print(f"\nTotal count: {len(fibonacci_numbers)}")
```

# Task 2
## write a program to simulate 1000 tosses of a fair coin (use np.random.binomial)
## Calculate the mean and standard deviation of that sample

```python
n_tosses = 1000
p_heads = 0.5
tosses = np.random.binomial(1, p_heads, size=n_tosses)

mean_tosses = np.mean(tosses)
std_tosses = np.std(tosses)

print(f"Number of tosses: {n_tosses}")
print(f"Number of heads: {np.sum(tosses)}")
print(f"Mean (proportion of heads): {mean_tosses:.4f}")
print(f"Standard deviation: {std_tosses:.4f}")
print(f"Expected mean: 0.5")
print(f"Expected std: {np.sqrt(0.5 * 0.5 / n_tosses):.4f}")
```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
n_points = 200
x = np.random.normal(loc=0, scale=5, size=n_points)
e = np.random.normal(loc=0, scale=1, size=n_points)

y = 0.5 * x + e

plt.figure(figsize=(8, 6))
plt.scatter(x, y, alpha=0.6, s=30)
plt.xlabel('x ~ N(0, 5)')
plt.ylabel('y = 0.5*x + e')
plt.title('Scatterplot of y vs x with Gaussian noise')
plt.grid(True, alpha=0.3)

x_line = np.linspace(min(x), max(x), 100)
y_line = 0.5 * x_line
plt.plot(x_line, y_line, 'r--', label='y = 0.5x (true relationship)')
plt.legend()
plt.show()
```
