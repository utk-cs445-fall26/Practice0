# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
# start with first two numbers
a = 0
b = 1

# continue generating numbers up to 1000000
while a <= 1000000:
    print(a)

    # calculate next two values in the sequence
    a, b = b, a + b
```

# Task 2
## write a program to simulate 1000 tosses of a fair coin (use np.random.binomial)
## Calculate the mean and standard deviation of that sample

```python
# 1 = heads
# 0 = tails
tosses = np.random.binomial(1, 0.5, 1000)

# mean
mean = np.mean(tosses)

# standard deviation
std = np.std(tosses)

print("Mean:", mean)
print("Standard deviation:", std)

```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
import matplotlib.pyplot as plt

# generate 100 x values from Gaussian (0, 5)
x = np.random.normal(0, 5, 100)

# generate 100 x random error values from Gaussian (0, 1)
e = np.random.normal(0, 1, 100)

# calculate y with y = 0.5*x + e
y = 0.5 * x + e

# Create x, y scatterplot
plt.scatter(x, y)

plt.xlabel("x")
plt.ylabel("y")

plt.show
```
