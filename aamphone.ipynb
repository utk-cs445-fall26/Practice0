# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
def fibonacci_up_to(limit):
    fibs = []
    # a = current fib num
    # b = the next one in line
    a, b = 0,1
    while a <= limit:
        fibs.append(a)
        # new a = old b, new b = old a + old b
        a, b = b, a + b
    return fibs

def main():
    limit = 1_000_000

    # Generate all Fibonacci numbers up to the limit
    result = fibonacci_up_to(limit)

    print(f"Fibonacci numbers up to {limit}:")
    for n in result:
        print(n)


main()
```

# Task 2
## write a program to simulate 1000 tosses of a fair coin (use np.random.binomial)
## Calculate the mean and standard deviation of that sample

```python
import numpy as np


# n = 1
# p = 0.5 probability - fair coin
# tossed 1,000 times
individual_flips = np.random.binomial(n=1, p=0.5, size=1000)

# calculate mean
mean_of_flips = np.mean(individual_flips)

# calculate standard deviation
std_of_flips = np.std(individual_flips)

print(f"Mean of sample: {mean_of_flips}")
print(f"Standard Deviation of Sample: {std_of_flips}")


```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
import numpy as np
import matplotlib.pyplot as plt

n_samples = 500

# generate x values from normal dist
x = np.random.normal(loc=0.0, scale=5, size=n_samples)

# generate random noise from normal dist
e = np.random.normal(loc=0.0, scale=1, size=n_samples)

y = 0.5*x+e

# set figure and scatter
plt.figure(figsize=(8, 6))
plt.scatter(x, y, alpha=0.6, color='blue', edgecolors='w', label='Data Points')

# generate x values to draw a smooth line reference
x_line = np.linspace(min(x), max(x), 100)
y_line = 0.5 * x_line
plt.plot(x_line, y_line, color='red', linestyle='--', linewidth=2, label='y = 0.5 * x')

# chart styling
plt.title('Scatterplot of $y = 0.5x + e$', fontsize=14)
plt.xlabel('x ~ N(0, 5)', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.axhline(0, color='black', linewidth=0.5, linestyle=':')
plt.axvline(0, color='black', linewidth=0.5, linestyle=':')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

# display the plot
plt.show()

```
