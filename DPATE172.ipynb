## Calculate the mean and standard deviation of that sample

```python
coin_tosses = np.random.binomial(1, 0.5, 1000)

print("Mean:", np.mean(coin_tosses))
print("Standard deviation:", np.std(coin_tosses))
```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
np.random.seed(445)
sample_size = 1000
x = np.random.normal(loc=0, scale=5, size=sample_size)
e = np.random.normal(loc=0, scale=1, size=sample_size)
y = 0.5 * x + e

plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.6, s=20, label=r"$y = 0.5x + e$")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Relationship with Gaussian Noise")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```
