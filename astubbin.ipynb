# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
def fibonacci(stop):
    numbers = [0, 1]
    while True:
        next_num = numbers[-1] + numbers[-2]
        if next_num > stop:
            break
        numbers.append(next_num)
    return numbers

print(fibonacci(1000000))
```

```python
samples = np.random.binomial(n=1, p=0.5, size=1000)
print("Mean:", np.mean(samples))
print("Std Dev:", np.std(samples))
```

## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
x = np.random.normal(loc=0.0, scale=5.0, size=500)
e = np.random.normal(loc=0.0, scale=1.0, size=500)
y = 0.5 * x + e
plt.scatter(x, y)
plt.title("y = 0.5*x + e")
plt.xlabel("x ~ Gaussian(0, 5)")
plt.ylabel("y")
```
