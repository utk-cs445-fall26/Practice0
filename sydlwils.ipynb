# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
a, b = 0, 1

print("Fibonacci sequence:")
while a <= 1000000:
    print(a)
    a, b = b, a + b
```

# Task 2
## write a program to simulate 1000 tosses of a fair coin (use np.random.binomial)
## Calculate the mean and standard deviation of that sample

```python
coin = np.random.binomial(1, .5, size=1000)

print("Mean: " + str(np.mean(coin)))
print("Standard Deviation: " + str(np.std(coin)))
```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
x = np.random.normal(0, 5, 300)
e = np.random.normal(0, 1, 300)

y = 0.5*x + e

plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Task 3 Scatterplot - Gaussian Distributions")
```
