# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
def fibonacci_up_to_million(million):
    fib_sequence = [0, 1]

    while True:
        next_num = fib_sequence[-1] + fib_sequence[-2]
        if next_num > 1000000:
            break
        fib_sequence.append(next_num)

    return fib_sequence

million = 1000000
fib_list = fibonacci_up_to_million(million)
print(fib_list)
```

# Task 2
## write a program to simulate 1000 tosses of a fair coin (use np.random.binomial)
## Calculate the mean and standard deviation of that sample

```python
tosses = np.random.binomial(1, 0.5, 1000)

mean = np.mean(tosses)
std = np.std(tosses)

print(f"Mean: {mean}")
print(f"Standard Deviation: {std}")
```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
x = np.random.normal(0, 5, 100) # 100 data points
e = np.random.normal(0, 1, 100)

y = 0.5 * x + e

plt.scatter(x, y, alpha=0.5, color="blue", edgecolors="none")
plt.title("Scatterplot of y = 0.5*x + e")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
