# Task 1
## Write a program to produce Fibonacci numbers up to 1000000

````python
fib = np.array([0, 1])

while fib[-1] + fib[-2] <= 1000000:
    fib = np.append(fib, fib[-1] + fib[-2])

print(fib)
````

# Task 2
## Write a program to simulate 1000 tosses of a fair coin (use `np.random.binomial`)
## Calculate the mean and standard deviation of that sample

````python
# Set random seed for reproducibility
np.random.seed(42)

# Simulate 1000 tosses of a fair coin
toss = np.random.binomial(1000, .5, size=1000)

# Print out the mean and standard deviation of the tosses
print(f"Mean: {toss.mean()}")
print(f"Standard deviation: {toss.std()}")
````

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use `numpy.random.normal` to generate gaussian distribution

````python
n = 1000

x = np.random.normal(0, 5, n)
e = np.random.normal(0, 1, n)
y = 0.5 * x + e

plt.scatter(x, y)

# Label axes and title
plt.title("Scatterplot of y = 0.5*x + e")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
````
