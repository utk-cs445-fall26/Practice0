# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
# Initialize the first two Fibonacci numbers and print the first one
num1 = 0
num2 = 1
print(num1)

# Print Fibonacci numbers until the first number greater than 1,000,000
while True:
    num1, num2 = num2, num1 + num2
    if num1 > 1000000:
        break
    print(num1)
```

# Task 2
## write a program to simulate 1000 tosses of a fair coin (use np.random.binomial)
## Calculate the mean and standard deviation of that sample

```python
# Initialize an empty list to store the results of the coin flips
coin_flips = []

# Simulate 1000 coin flips and store the results in the list
for _ in range(1000):
    flip = np.random.binomial(1, 0.5)
    coin_flips.append(flip)

# Calculate and print the mean and standard deviation of the coin flips
print(f"Mean of coin flips: {np.mean(coin_flips)}")
print(f"Standard deviation of coin flips: {np.std(coin_flips)}")
```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
# Generate random data for x and e
x = np.random.normal(0, 5, 1000)
e = np.random.normal(0, 1, 1000)

# Initialize y as a linear function of x with some added noise e
y = 0.5 * x + e

# Create a scatterplot of the data
plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Scatterplot of y = 0.5*x + e")
plt.show()
```
